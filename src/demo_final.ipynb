# Demo final – 5 noticias del *test set*

Como prueba final del proyecto, hemos querido completar la tarea que nos propusimos en un inicio. Tratar de sacar "tickets" de las noticias, o en lo que ha quedado más, un pequeño informe, que contiene un resumen, hemos extraido entidades y clasificado la noticia en una de las 5 categorias que tenemos. Para ello este notebook ejecuta el **pipeline completo** sobre 5 noticias del conjunto de test seleccionadas aleatoriamente:

1. **Clasificación temática** con **FinBERT fine-tuneado (5 topics)**
2. **Summarization** con `falconsai/text_summarization`
3. **NER** con `boltuix/NeuroBERT-NER`

Para cada noticia se muestra el **texto principal**  para mejorar la interpretabilidad y evaluar si verdaderamente está funcionando, debajo, un bloque tipo *ticker* con:
**Resumen**, **Entidades**, **Topic predicho** y si la predicción es **correcta**.


```python
# Imports
import os
import pandas as pd
import numpy as np
import torch
import matplotlib.pyplot as plt

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    AutoModelForSeq2SeqLM,
    AutoModelForTokenClassification,
    AutoConfig,
)

from IPython.display import display, Markdown, HTML

try:
    import joblib
except ImportError:
    joblib = None
```

## Rutas y configuración

```python
import os
print("CWD:", os.getcwd())

TEST_CSV = "../data/definitivos/splits/test.csv"
MODEL_DIR = "./Segunda_Resolucion/Text_classification/models/finbert_topicclf"

# Label encoder guardado junto al modelo (recomendado)
LABEL_ENCODER_PATH = os.path.join(MODEL_DIR, "label_encoder.joblib")

# Modelos (de vuestros notebooks)
SUM_MODEL_NAME = "falconsai/text_summarization"
NER_MODEL_NAME = "boltuix/NeuroBERT-NER"

# Demo (solo vamos a usar 5 samples osea que bien así)
N_SAMPLES = 5
RANDOM_STATE = 42

# Inferencia
MAX_LEN_CLS = 512
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("DEVICE:", DEVICE)
```

## Cargar dataset

```python
df_test = pd.read_csv(TEST_CSV)

# Crear text_input si no existe
if "text_input" not in df_test.columns:
    df_test["text_input"] = df_test["headline"].fillna("") + " " + df_test["summary"].fillna("")

# Texto principal: preferimos article_text si existe
TEXT_COL = "article_text" if "article_text" in df_test.columns else "text_input"

# samples
samples = df_test.sample(n=min(N_SAMPLES, len(df_test)), random_state=RANDOM_STATE).reset_index(drop=True)

display(samples[[c for c in ["headline", "topic", "label"] if c in samples.columns]].head())
print("Texto principal:", TEXT_COL)
```

## Cargar modelos (clasificación, resumen, NER)

En este caso, y como queremos que salga lo mejor posible, hemos escogido los mejores modelos de cada task. Recogidos también al inicio del notebook

```python
# =========================
# 1) Clasificación (FinBERT fine-tuneado)
# =========================
cls_tokenizer = AutoTokenizer.from_pretrained(MODEL_DIR)
cls_model = AutoModelForSequenceClassification.from_pretrained(MODEL_DIR).to(DEVICE)
cls_model.eval()

if joblib is not None and os.path.exists(LABEL_ENCODER_PATH):
    le = joblib.load(LABEL_ENCODER_PATH)
    class_names = list(le.classes_)
else:
    n_classes = int(df_test["label"].max()) + 1 if "label" in df_test.columns else 5
    class_names = [str(i) for i in range(n_classes)]

print("Clases:", class_names)
print("num_labels (modelo):", cls_model.config.num_labels)

# =========================
# 2) Summarization
# =========================
sum_tokenizer = AutoTokenizer.from_pretrained(SUM_MODEL_NAME)
sum_model = AutoModelForSeq2SeqLM.from_pretrained(SUM_MODEL_NAME).to(DEVICE)
sum_model.eval()

# =========================
# 3) NER
# =========================
ner_config = AutoConfig.from_pretrained(NER_MODEL_NAME)
id2label = ner_config.id2label
label2id = ner_config.label2id

ner_tokenizer = AutoTokenizer.from_pretrained(NER_MODEL_NAME)
ner_model = AutoModelForTokenClassification.from_pretrained(
    NER_MODEL_NAME,
    num_labels=len(id2label),
    id2label=id2label,
    label2id=label2id
).to(DEVICE)
ner_model.eval()

print("NER labels:", len(id2label))
```

## Funciones: clasificación, summarization y NER

```python
def predict_topic(text: str):
    enc = cls_tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        padding=True,
        max_length=MAX_LEN_CLS
    )
    enc = {k: v.to(DEVICE) for k, v in enc.items()}
    with torch.no_grad():
        logits = cls_model(**enc).logits
        probs = torch.softmax(logits, dim=-1).squeeze(0).detach().cpu().numpy()
    pred_id = int(np.argmax(probs))
    pred_name = class_names[pred_id] if pred_id < len(class_names) else str(pred_id)
    return pred_id, pred_name, probs

# ha habido que bajar el max_new_tokens porque iba lentisimo(no he conseguido averiguar porque)
def summarize_text(text: str,
                   max_new_tokens: int = 96,
                   min_new_tokens: int = 24,
                   num_beams: int = 4):
    """Resumen con falconsai/text_summarization.
    Nota: Si el texto es muy largo, se truncará a lo que el modelo soporte.
    """
    if not text or not text.strip():
        return ""

    # Tokenización (truncation=True para evitar errores con textos largos)
    enc = sum_tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=1024
    ).to(DEVICE)

    with torch.no_grad():
        gen_ids = sum_model.generate(
            **enc,
            max_new_tokens=max_new_tokens, 
            min_new_tokens=min_new_tokens,
            num_beams=num_beams,
            length_penalty=1.0,
            no_repeat_ngram_size=3,
            early_stopping=True,
        )

    summary = sum_tokenizer.decode(gen_ids[0], skip_special_tokens=True).strip()
    return summary


def extract_entities(text: str, max_len: int = 512, top_k: int = 12):
    """NER con boltuix/NeuroBERT-NER.
    Devuelve lista de (LABEL, span_text) ya agrupada por BIO.
    """
    if not text or not text.strip():
        return []

    enc = ner_tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=max_len,
        padding=True
    )
    enc = {k: v.to(DEVICE) for k, v in enc.items()}

    with torch.no_grad():
        logits = ner_model(**enc).logits  # [B, T, C]
        pred_ids = torch.argmax(logits, dim=-1)[0].detach().cpu().numpy().tolist()

    # Tokens del modelo
    tokens = ner_tokenizer.convert_ids_to_tokens(enc["input_ids"][0].detach().cpu().numpy().tolist())
    tags = [id2label[i] for i in pred_ids]

    # Agrupar BIO en spans y reconstruir texto aproximado(prueba que mejor ha funcionado, las demas un desastre) (dejarlo así es la que mejor pinta tiene)
    entities = []
    current_label = None
    current_tokens = []

    def flush():
        nonlocal current_label, current_tokens
        if current_label and current_tokens:
            # Reconstrucción subword -> texto
            span = ner_tokenizer.convert_tokens_to_string(current_tokens).strip()
            if span:
                entities.append((current_label, span))
        current_label = None
        current_tokens = []

    for tok, tag in zip(tokens, tags):
        if tok in (ner_tokenizer.cls_token, ner_tokenizer.sep_token, ner_tokenizer.pad_token):
            continue

        if tag.startswith("B-"):
            flush()
            current_label = tag[2:]
            current_tokens = [tok]
        elif tag.startswith("I-") and current_label == tag[2:]:
            current_tokens.append(tok)
        else:
            flush()

    flush()

    # Filtrar duplicados simples y limitar top_k
    # (muchas veces se repiten spans muy pequeños)
    seen = set()
    cleaned = []
    for lab, span in entities:
        key = (lab, span.lower())
        if key not in seen:
            cleaned.append((lab, span))
            seen.add(key)

    return cleaned[:top_k]
```

## Visualización de los "informes"/"tickers" 

```python
def render_ticker(summary: str, entities, pred_topic: str, is_correct: bool | None):
    correct_text = "N/A" if is_correct is None else ("✅ Correcta" if is_correct else "❌ Incorrecta")
    correct_style = "color:#2e7d32;font-weight:700;" if is_correct else "color:#c62828;font-weight:700;"
    if is_correct is None:
        correct_style = "color:#616161;font-weight:700;"

    if isinstance(entities, (list, tuple)):
        entities_str = ", ".join([f"{lab}: {span}" for lab, span in entities]) if entities else "-"
    else:
        entities_str = str(entities)

    summary = summary if summary else "-"

    html = f"""
    <div style="border:1px solid #ddd;border-radius:10px;padding:12px;margin-top:10px;">
      <div style="display:flex;gap:14px;flex-wrap:wrap;">
        <div style="flex:1;min-width:260px;">
          <div style="font-size:12px;color:#555;">Resumen</div>
          <div style="font-size:14px;">{summary}</div>
        </div>
        <div style="flex:1.2;min-width:320px;">
          <div style="font-size:12px;color:#555;">Entidades (NER)</div>
          <div style="font-size:14px;">{entities_str}</div>
        </div>
        <div style="flex:0.8;min-width:220px;">
          <div style="font-size:12px;color:#555;">Clasificación (Topic)</div>
          <div style="font-size:14px;font-weight:700;">{pred_topic}</div>
        </div>
        <div style="flex:0.6;min-width:160px;text-align:right;">
          <div style="font-size:12px;color:#555;">¿Correcta?</div>
          <div style="font-size:14px;{correct_style}">{correct_text}</div>
        </div>
      </div>
    </div>
    """

    return HTML(html)
```

## Ejecutar demo (5 noticias)

```python
    def top_k(probs, k=3):
        idx = np.argsort(probs)[::-1][:k]
        return [(class_names[i], float(probs[i])) for i in idx]

    for i, row in samples.iterrows():
        headline = str(row.get("headline", f"Noticia {i+1}"))
        true_topic_name = row.get("topic", None)
        true_label = row.get("label", None)

        text = str(row.get(TEXT_COL, "")) or str(row.get("text_input", ""))
        text = text.strip()

        # Predicción clasificación
        pred_id, pred_topic, probs = predict_topic(text)

        # Correctitud (si tenemos GT)
        is_correct = None
        if true_topic_name is not None and isinstance(true_topic_name, str) and true_topic_name.strip():
            is_correct = (pred_topic == true_topic_name)
        elif true_label is not None and not pd.isna(true_label):
            is_correct = (pred_id == int(true_label))

        # Summarization + NER
        summary = summarize_text(text)
        entities = extract_entities(text)

        # Render
        display(Markdown(f"""---
## {i+1}) {headline}
"""))
        display(Markdown(f"""**Texto principal ({TEXT_COL})**"""))
        display(Markdown(text))

        display(render_ticker(summary=summary, entities=entities, pred_topic=pred_topic, is_correct=is_correct))
        display(Markdown(f"**Top-3 topics:** {top_k(probs, 3)}")) 
```

## Conclusion Final

En estos ejemplos que hemos probado, podemos observar que en 3/5 de los casos se asigna correctamente la categoría, y nos hemos dado cuenta de que esto ocurre sobre todo en textos más concretos y claramente enfocados en noticias financieras, con un énfasis claro en resultados empresariales, crecimiento corporativo o evolución de mercado. Para este tipo de noticias el modelo no suele tener problemas en asignar la categoría correctamente. Los fallos aparecen principalmente entre algunos tópicos cercanos y están muy relacionados con las clases que tienen menos noticias, tal y como se explica en el informe.

En cuanto a la generación de resúmenes, hemos visto que existen limitaciones puntuales de coherencia en los textos generados. Creemos que estas limitaciones están en gran parte derivadas de las restricciones de longitud de los modelos Transformer y de la necesidad de truncar textos largos. Por ejemplo, en el cuarto caso aparece una secuencia de varias comas consecutivas (“,,, ”) que no tiene sentido viendo el texto original, lo que indica pérdidas puntuales de coherencia durante la generación.

De forma similar, el módulo de reconocimiento de entidades presenta errores asociados a la tokenización por subpalabras. Esto se puede observar claramente en ejemplos como Walmart, donde la entidad aparece fragmentada en múltiples tokens (ORG: wal, ORG: ##mart, ORG: ##mt, ORG: ##mart inc.), o en noticias relacionadas con Oracle, donde se detectan fragmentaciones como ORG: oracle, ORG: ##cl, ORG: ##ci, o ORG: oracle cloud. Este comportamiento explica también la aparición de cantidades monetarias y porcentajes fragmentados, como $ 3. 3 billion o 28 %, y es una limitación directa del enfoque de tokenización utilizado.

Tras analizar los distintos outputs, hemos decidido incluir en la documentación final el caso de Oracle, ya que es el que mejor representa el comportamiento esperado del sistema. En este ejemplo, la clasificación temática es coherente, las entidades principales se identifican de una forma razonable aunque siguen existiendo errores por subpalabras, y el resumen es el que más nos convence en términos de coherencia general y conservación del contexto financiero, motivo por el cual nos hemos decantado por este caso como ejemplo final.
